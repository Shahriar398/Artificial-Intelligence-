**Local Beam Search**

```python
# Node values
values = {
    'S': 1,
    'A': 3, 'B': 5, 'C': 4,
    'D': 2, 'E': 8, 'F': 6, 'G': 7, 'H': 9
}

# Tree connections
tree = {
    'S': ['A', 'B', 'C'],
    'A': ['D', 'E'],
    'B': ['F'],
    'C': ['G', 'H'],
    'D': [], 'E': [], 'F': [], 'G': [], 'H': []
}

def local_beam_search(start, k=2):
    states = [start]

    while True:
        all_neighbors = []

        for state in states:
            all_neighbors.extend(tree[state])

        if not all_neighbors:
            break

        states = sorted(all_neighbors, key=lambda x: values[x], reverse=True)[:k]

    best = max(states, key=lambda x: values[x])
    return best, values[best]

result = local_beam_search('S')
print("Local Beam Search Result:", result)
```
