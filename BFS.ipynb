# **BFS**

```python
Ex1 = {
    'S': {'A':3, 'C':2 },
    'A': {'B': 1, 'C': 1},
    'C': {'D':4},
    'B': {'G':2},
    'D': {'G':2},
    'G': {}
}
```

```python
list (Ex1['S'].keys())

list (Ex1.keys())
```

```python
def bfs(startingNode, destinationNode, G):

    reached = []
    expand_sequence = []
    frontier = []

    snode = {
        'S': startingNode,
        'Path cost': 0,
        'Path': [startingNode]
    }

    frontier.append(snode)
    reached.append(startingNode)

    while len(frontier) > 0:

        unode = frontier.pop(0)
        u = unode['S']

        expand_sequence.append(u)

        # Goal Test
        if u == destinationNode:
            return {
                'Path': unode['Path'],
                'Cost': unode['Path cost'],
                'Expanded': expand_sequence
            }

        # Expand neighbors
        for v in G[u]:
            if v not in reached:
                vnode = {
                    'S': v,
                    'Path cost': unode['Path cost'] + G[u][v],
                    'Path': unode['Path'] + [v]
                }

                frontier.append(vnode)
                reached.append(v)

    return "Goal not found"
```

```python
result = bfs('S', 'G', Ex1)
result
```
