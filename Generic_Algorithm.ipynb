```python
import random
import numpy as np

# 1. Objective Function
def func(x):
    return (x**2)/2 - 3*x


# 2. Initialize Population (5 bit, 4 population)
def initPopulation(b=5, n=4):
    return [random.choices([0, 1], k=b) for _ in range(n)]


# 3. Binary to Decimal
def binToDecimal(binList):
    dec = 0
    power = len(binList)
    for bit in binList:
        dec += bit * 2**(power - 1)
        power -= 1
    return dec


# 4. Fitness Function
def getFitness(population):
    fitness = []
    for sol in population:
        x = binToDecimal(sol)
        val = func(x)
        fitness.append(max(0, val))
    return fitness


# 5. Selection (Roulette Wheel)
def select_parent(pop):
    fitness = getFitness(pop)
    total = sum(fitness)
    prob = [f/total for f in fitness]
    cumProb = np.cumsum(prob)

    parents = []
    for _ in range(4):
        r = random.random()
        for i in range(len(cumProb)):
            if r <= cumProb[i]:
                parents.append(pop[i])
                break
    return parents


# 6. Crossover (fixed point = 3)
def cross(parents):
    off = []
    for i in range(0, len(parents), 2):
        x = parents[i]
        y = parents[i+1]

        cp = 3  # fixed crossover point

        off1 = x[:cp] + y[cp:]
        off2 = y[:cp] + x[cp:]

        off.append(off1)
        off.append(off2)
    return off


# 7. Mutation (ONLY 4th BIT → index 3)
def mutate(offsprings):
    for child in offsprings:
        child[3] = 1 - child[3]   # flip 4th bit
    return offsprings


# 8. Best Solution
def best(pop):
    fit = getFitness(pop)
    idx = fit.index(max(fit))
    return pop[idx], max(fit)


# 9. Main GA Loop
population = initPopulation(5, 4)

for gen in range(3):
    print(f"\nGeneration {gen+1}")
    print("Best:", best(population))

    parents = select_parent(population)
    offsprings = cross(parents)
    offsprings = mutate(offsprings)

    population = population + offsprings
```

```python
import random
import numpy as np

# 1. Objective Function
def func(x):
    return (x**2)/2 - 3*x


# 2. Initialize Population (5 bit, 4 population)
def initPopulation(b=5, n=4):
    return [random.choices([0, 1], k=b) for _ in range(n)]


# 3. Binary to Decimal
def binToDecimal(binList):
    dec = 0
    power = len(binList)
    for bit in binList:
        dec += bit * 2**(power - 1)
        power -= 1
    return dec


# 4. Fitness Function
def getFitness(population):
    fitness = []
    for sol in population:
        x = binToDecimal(sol)
        val = func(x)
        fitness.append(max(0, val))
    return fitness


# 5. Selection (Roulette Wheel)
def select_parent(pop):
    fitness = getFitness(pop)
    total = sum(fitness)

    # avoid division by zero
    if total == 0:
        prob = [1/len(fitness)] * len(fitness)
    else:
        prob = [f/total for f in fitness]

    cumProb = np.cumsum(prob)

    parents = []
    for _ in range(4):
        r = random.random()
        for i in range(len(cumProb)):
            if r <= cumProb[i]:
                parents.append(pop[i])
                break
    return parents


# 6. Crossover (fixed point = 3)
def cross(parents):
    off = []
    for i in range(0, len(parents), 2):
        x = parents[i]
        y = parents[i+1]

        cp = 3

        off1 = x[:cp] + y[cp:]
        off2 = y[:cp] + x[cp:]

        off.append(off1)
        off.append(off2)
    return off


# 7. Mutation (ONLY 4th BIT → index 3)
def mutate(offsprings):
    for child in offsprings:
        child[3] = 1 - child[3]
    return offsprings


# 8. Best Solution
def best(pop):
    fit = getFitness(pop)
    idx = fit.index(max(fit))
    return pop[idx], max(fit)


# 9. MAIN (ONLY 1 GENERATION)
population = initPopulation(5, 4)

print("Initial Population:")
print(population)
print("Best:", best(population))

#  ONLY ONE GENERATION RUN
parents = select_parent(population)
offsprings = cross(parents)
offsprings = mutate(offsprings)

population = population + offsprings

print("\nAfter 1 Generation:")
print(population)
print("Best:", best(population))
```
