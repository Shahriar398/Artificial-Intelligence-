```python
romanian_map = {
    'Arad': {'Sibiu':140, 'Zerind':75, 'Timisoara':118},
    'Zerind': {'Arad':75, 'Oradea':71},
    'Oradea': {'Zerind':71, 'Sibiu': 151},
    'Sibiu': {'Arad':140, 'Oradea':151, 'Fagaras':99, 'Rimnicu':80},
    'Timisoara': {'Arad': 118, 'Lugoj':111},
    'Lugoj': {'Timisoara':111, 'Mehadia':70},
    'Mehadia': {'Lugoj': 70, 'Drobeta':75},
    'Drobeta': {'Mehadia':75, 'Craiova':120},
    'Craiova': {'Drobeta': 120, 'Rimnicu': 146, 'Pitesti': 138},
    'Rimnicu': {'Sibiu': 80, 'Craiova': 146, 'Pitesti': 97},
    'Fagaras': {'Sibiu':99, 'Bucharest':211},
    'Pitesti': {'Rimnicu': 97, 'Craiova':138, 'Bucharest':101},
    'Bucharest': {'Fagaras':211, 'Pitesti': 101, 'Giurgiu':90, 'Urziceni':85},
    'Giurgiu': {'Bucharest': 90},
    'Urziceni': {'Bucharest': 85, 'Vaslui':142, 'Hirsova':98},
    'Hirsova': {'Urziceni':98, 'Eforie':86},
    'Eforie': {'Hirsova':86},
    'Vaslui': {'Iasi': 92, 'Urziceni':142},
    'Iasi': {'Vaslui':92, 'Neamt': 87},
    'Neamt': {'Iasi':87}
}
```

```python
def bfs(startingNode, destinationNode, G):
    reached = [] # to store reached (generated) nodes
    expand_sequence = [] # to store the order of nodes expanded
    frontier = []

    snode = {'City':startingNode,'Path cost':0, 'Path':[startingNode]}
    frontier.append(snode)    # insert the starting node
    reached.append(startingNode)

    while len(frontier) > 0:
        print(frontier)
        unode  = frontier.pop(0)     # selceting the node to expand
        u = unode['City']
        expand_sequence.append(u)
        print(u + " expanding..")
        # expand u

        for v in G[u].keys():
            if v not in reached: # process if v is not reached (generated) yet
                print(v+" generated.")
                cost = unode['Path cost'] + G[u][v]
                path =  unode['Path']+ [v]
                vnode = {'City': v,'Path cost': cost,'Path': path}
                frontier.append(vnode)# store the new child node in the frontier
                reached.append(v)
                # goal test
                if v == destinationNode:
                    #print(vnode)
                    print(expand_sequence)
                    return vnode # breaking from while
    print('Failed')
```

```python
bfs('Arad', 'Bucharest', romanian_map)
```

DFS





```python
def dfs(startingNode, destinationNode, G):

    reached = []
    expand_sequence = []
    frontier = []

    snode = {'City': startingNode, 'Path cost': 0, 'Path': [startingNode]}
    frontier.append(snode)
    reached.append(startingNode)

    while len(frontier) > 0:

        print(frontier)
        unode = frontier.pop()   # stack pop
        u = unode['City']

        expand_sequence.append(u)
        print(u + " expanding..")

        if u == destinationNode:
            print(expand_sequence)
            return unode

        for v in G[u].keys():
            if v not in reached:

                print(v + " generated.")

                cost = unode['Path cost'] + G[u][v]
                path = unode['Path'] + [v]

                vnode = {'City': v, 'Path cost': cost, 'Path': path}

                frontier.append(vnode)
                reached.append(v)

    print("Failed")
```

```python
dfs('Arad', 'Bucharest', romanian_map)
```

UCS


```python
EX1 = {
    'S': {'A': 1,'B': 2,'C': 1},
    'A':{'C':1,'D':2,'G':6},
    'B': {'D':2,'E':3},
    'C':{'G':8},
    'D':{'E':1},
    'E':{'G':1},
    'G':{}
}
```

```python
from queue import PriorityQueue


def ucs(snode, dnode, GRAPH):
    expanded = [] # to store expanded nodes
    frontier = PriorityQueue()

    node = (0,snode,[snode]) #(g(n) = path cost, city name, path)
    frontier.put(node)
    i = 1
    while not frontier.empty():
        node = frontier.get() # select a node for expansion
        u = node[1]
        if u in expanded:
          continue
        print()
        print(u+ " selected. Cost: "+ str(node[0]))
        if u == dnode:# goal test
            print("*** GOAL TEST SATISFIED ***")
            print()
            return node # path
        expanded.append(u)
        print('['+str(i)+']: '+u+" expanding...")
        i = i +1
        for v in GRAPH[u].keys():#expanding
            if v not in expanded:
                g = node[0] + GRAPH[u][v]
                path = node[2] + [v]
                frontier.put((g,v,path))
                print(u + "->"+v + ", Path: "+ str(path) + ", Cost:"+ str(g))
        #print(frontier.queue)
    print('Failed')
```

```python
ucs('S', 'G', EX1)
```

GBFS


```python
h1 = {'S':10,'A':8, 'B':5, 'C':5,
     'D':4,'E':2, 'G':0}
```

```python
from queue import PriorityQueue

def gbfs(snode, dnode, GRAPH, heur):
    expanded = [] # to store expanded nodes
    frontier = PriorityQueue()

    snode = (heur[snode],snode,[snode],0) # (h, city, path, path cost)
    frontier.put(snode)
    i =1
    while not frontier.empty():
        node = frontier.get()
        u = node[1]
        if u in expanded:
          continue

        print(u+ " selected. Cost: "+ str(node[0]))
        if u== dnode:#Goal Test
            print(node)
            print(expanded)
            return node #path
        expanded.append(u)
        print('['+str(i)+']: '+u+" expanding...")
        i = i+1
        for v in GRAPH[u].keys():
            if v not in expanded:
                g = node[3] + GRAPH[u][v]
                path = node[2]+[v]
                frontier.put((heur[v],v,path, g))
                print(u + "->"+v + ", Path: "+ str(path) + ", h:"+ str(heur[v]))
    print('Failed')
```

```python
h1 = {'S':10,'A':8, 'B':5, 'C':5,
     'D':4,'E':2, 'G':0}
```

```python
from queue import PriorityQueue

def astar(snode, dnode, GRAPH, heur):

    expanded = []
    frontier = PriorityQueue()

    frontier.put((heur[snode], snode, [snode], 0))

    while not frontier.empty():

        node = frontier.get()
        u = node[1]

        if u in expanded:
            continue

        print(u,"selected. f =", node[0])

        if u == dnode:
            print(node)
            return node

        expanded.append(u)

        for v in GRAPH[u].keys():

            if v not in expanded:

                g = node[3] + GRAPH[u][v]
                f = g + heur[v]
                path = node[2] + [v]

                frontier.put((f,v,path,g))
```
